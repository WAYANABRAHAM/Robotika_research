```python
import serial
import time  # Tambahkan ini

# Buka port
arduinoData = serial.Serial('COM5', 115200, timeout=0)
time.sleep(2)  # Beri jeda 2 detik agar Arduino siap

while True:
    myData = input('Kirimkan perintah (1/0): ')  
    if myData == "1":
        arduinoData.write(b'1')
    elif myData == "0":  
        arduinoData.write(b'0')
```

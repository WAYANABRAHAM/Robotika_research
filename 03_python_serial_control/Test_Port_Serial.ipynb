# Test Port Serial — Motor ON/OFF

Contoh ini cocok dengan firmware basic yang menerima karakter `1` dan `0` pada 115200 baud. Seluruh koneksi berada di dalam fungsi agar notebook tidak membuka port atau menggerakkan motor saat **Run All**. Pastikan area motor aman dan tutup Serial Monitor sebelum memanggil fungsi.

```python
import time
import serial

def kontrol_motor(port: str, baudrate: int = 115200) -> None:
    with serial.Serial(port, baudrate, timeout=1) as esp:
        time.sleep(2)  # ESP32 dapat reset saat port dibuka
        try:
            esp.write(b'0')  # mulai pada keadaan OFF
            print('Port siap. Perintah: 1=ON, 0=OFF, q=keluar')
            while True:
                command = input('Perintah: ').strip().lower()
                if command == 'q':
                    break
                if command in ('1', '0'):
                    esp.write(command.encode('ascii'))
                    esp.flush()
                else:
                    print('Masukkan 1, 0, atau q')
        finally:
            esp.write(b'0')  # minta firmware berhenti saat keluar
            esp.flush()
```
